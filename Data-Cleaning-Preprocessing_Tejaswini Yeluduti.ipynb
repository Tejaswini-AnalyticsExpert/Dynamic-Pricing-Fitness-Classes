**1. Import Required Libraries**

```python
# =========================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# =========================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings('ignore')
```

**2. Upload & Load Datasets**

```python
# =========================================================
# STEP 2: UPLOAD & LOAD DATASETS
# =========================================================

from google.colab import files

uploaded = files.upload()

# Load datasets
df1 = pd.read_csv('Classes April-May 2018.csv')
df2 = pd.read_csv('Classes June 2018.csv')

print("Datasets Loaded Successfully")
```

**3. Initial Data Inspection**

```python
# =========================================================
# STEP 3: INITIAL DATA INSPECTION
# =========================================================

print("April-May Dataset Shape:", df1.shape)
print("June Dataset Shape:", df2.shape)

print("\nFirst 5 Rows")
display(df1.head())

print("\nDataset Information")
print(df1.info())

print("\nMissing Values")
print(df1.isnull().sum())
```

**4. Merge Datasets**

```python
# =========================================================
# STEP 4: MERGE DATASETS
# =========================================================

df = pd.concat([df1, df2], ignore_index=True)

print("Combined Dataset Shape:", df.shape)
```

**5. Clean Column Names**

```python
# =========================================================
# STEP 5: CLEAN COLUMN NAMES
# =========================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace(r'[^a-zA-Z0-9_]', '', regex=True)
)

print(df.columns)
```

```python
# =========================================================
# STEP 5: CLEAN & STANDARDIZE COLUMN NAMES
# =========================================================

# Convert column names into professional standard format

df.columns = [
    'activity_site_id',
    'activity_description',
    'booking_end_datetime',
    'booking_start_time',
    'max_bookees',
    'number_booked',
    'price_inr'
]

print("Updated Professional Column Names:\n")
print(df.columns)
```

**6. Remove Duplicate Records**

```python
# =========================================================
# STEP 6: REMOVE DUPLICATE RECORDS
# =========================================================

duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

**7. Handle Missing Values**

```python
# =========================================================
# STEP 7: HANDLE MISSING VALUES
# =========================================================

print("Missing Values Before Treatment")
print(df.isnull().sum())

# Numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Treatment")
print(df.isnull().sum())
```

**8. Convert Data Types**

```python
# =========================================================
# STEP 8: CONVERT DATA TYPES
# =========================================================

date_col = 'booking_end_datetime'

df[date_col] = pd.to_datetime(
    df[date_col],
    errors='coerce'
)

print(df.dtypes)
```

**9. Feature Engineering**

```python
# =========================================================
# STEP 9: FEATURE ENGINEERING
# =========================================================

if date_col in df.columns:

    df['year'] = df[date_col].dt.year
    df['month'] = df[date_col].dt.month
    df['day'] = df[date_col].dt.day
    df['day_name'] = df[date_col].dt.day_name()

print("Feature Engineering Completed")
```

**10. Feature Scaling**

```python
# =========================================================
# STEP 10: FEATURE SCALING
# =========================================================

scale_cols = [
    'price_inr',
    'number_booked',
    'max_bookees'
]

scaler = StandardScaler()

df[scale_cols] = scaler.fit_transform(df[scale_cols])

print("Scaled Columns:")
print(scale_cols)
```

**11. Final Quality Check**

```python
# =========================================================
# STEP 11: FINAL DATA QUALITY CHECK
# =========================================================

print("Final Dataset Shape:", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Records")
print(df.duplicated().sum())

display(df.head())
```

**12. Export Cleaned Dataset**

```python
# =========================================================
# STEP 12: EXPORT CLEANED DATASET
# =========================================================

output_file = 'cleaned_preprocessed_dataset.csv'

df.to_csv(output_file, index=False)

print("Dataset Exported Successfully")

files.download(output_file)
```
